```python
# Import Path to manage file and folder paths
from pathlib import Path

# Import pandas to load and manipulate tabular data
import pandas as pd

# Import joblib to save trained models
import joblib

# Import Pipeline to combine preprocessing and model training
from sklearn.pipeline import Pipeline

# Import StandardScaler to standardize numerical features
from sklearn.preprocessing import StandardScaler

# Import the five required classification algorithms
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.svm import SVC

# Define the random seed for reproducibility
RANDOM_STATE = 42
```

```python
# Define the path of the processed training dataset
TRAIN_PATH = Path("../data/processed/train_data.csv")

# Define the path of the processed testing dataset
TEST_PATH = Path("../data/processed/test_data.csv")

# Load the training dataset into a DataFrame
train_data = pd.read_csv(TRAIN_PATH)

# Load the testing dataset into a DataFrame
test_data = pd.read_csv(TEST_PATH)

# Separate the input features from the target column
X_train = train_data.drop(columns=["price_range"])

# Store the training target values
y_train = train_data["price_range"]

# Separate the testing input features from the target column
X_test = test_data.drop(columns=["price_range"])

# Store the testing target values
y_test = test_data["price_range"]

# Display the dimensions of the data
print("Training features:", X_train.shape)
print("Training target:", y_train.shape)
print("Testing features:", X_test.shape)
print("Testing target:", y_test.shape)
```

```python
# Create a pipeline for Logistic Regression with feature scaling
logistic_regression_pipeline = Pipeline([
    # Standardize all input features before training
    ("scaler", StandardScaler()),

    # Train the Logistic Regression classifier
    ("model", LogisticRegression(
        max_iter=2000,
        random_state=RANDOM_STATE
    ))
])

# Create a pipeline for the Decision Tree classifier
decision_tree_pipeline = Pipeline([
    # Train the Decision Tree without feature scaling
    ("model", DecisionTreeClassifier(
        random_state=RANDOM_STATE
    ))
])

# Create a pipeline for the K-Nearest Neighbor classifier
knn_pipeline = Pipeline([
    # Standardize features because KNN is distance-based
    ("scaler", StandardScaler()),

    # Train the KNN classifier
    ("model", KNeighborsClassifier(
        n_neighbors=5
    ))
])

# Create a pipeline for Gaussian Naive Bayes
naive_bayes_pipeline = Pipeline([
    # Standardize the input features
    ("scaler", StandardScaler()),

    # Train the Gaussian Naive Bayes classifier
    ("model", GaussianNB())
])

# Create a pipeline for Support Vector Machine
svm_pipeline = Pipeline([
    # Standardize features because SVM is scale-sensitive
    ("scaler", StandardScaler()),

    # Train an SVM with probability estimates enabled
    ("model", SVC(
        probability=True,
        random_state=RANDOM_STATE
    ))
])
```

```python
# Store each model using a readable name
models = {
    "Logistic Regression": logistic_regression_pipeline,
    "Decision Tree": decision_tree_pipeline,
    "KNN": knn_pipeline,
    "Naive Bayes": naive_bayes_pipeline,
    "Support Vector Machine": svm_pipeline
}

# Display the names of the models
print("Models selected for training:")
for model_name in models:
    print("-", model_name)
```

```python
# Train each model using the training features and training labels
for model_name, model_pipeline in models.items():
    # Fit the complete preprocessing and model pipeline
    model_pipeline.fit(X_train, y_train)

    # Display a confirmation message after training
    print(f"{model_name} trained successfully.")
```

```python
# Define the folder where trained models will be stored
MODEL_DIR = Path("../model")

# Create the model folder if it does not already exist
MODEL_DIR.mkdir(parents=True, exist_ok=True)

# Define filenames for each trained model
model_filenames = {
    "Logistic Regression": "logistic_regression.joblib",
    "Decision Tree": "decision_tree.joblib",
    "KNN": "knn.joblib",
    "Naive Bayes": "naive_bayes.joblib",
    "Support Vector Machine": "svm.joblib"
}

# Save each trained pipeline to disk
for model_name, model_pipeline in models.items():
    # Create the complete output path
    model_path = MODEL_DIR / model_filenames[model_name]

    # Save the complete pipeline, including scaling where applicable
    joblib.dump(model_pipeline, model_path)

    # Display the saved model path
    print(f"{model_name} saved to: {model_path}")
```

```python
# List all files saved inside the model directory
saved_model_files = list(MODEL_DIR.glob("*.joblib"))

# Display the number of saved model files
print("Number of saved models:", len(saved_model_files))

# Display each saved model filename
for model_file in saved_model_files:
    print(model_file.name)
```
