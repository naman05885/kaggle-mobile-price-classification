```python
from pathlib import Path

import pandas as pd
from sklearn.model_selection import train_test_split

RANDOM_STATE = 42
TARGET_COLUMN = "price_range"

RAW_DATA_PATH = Path("../data/raw/train.csv")
PROCESSED_DIR = Path("../data/processed")

PROCESSED_DIR.mkdir(parents=True, exist_ok=True)

print("Configuration loaded successfully.")
```

```python
df = pd.read_csv(RAW_DATA_PATH)

print("Dataset shape:", df.shape)
display(df.head())
```

```python
print("Missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())
print("Target column:", TARGET_COLUMN)
print("Target classes:")
display(df[TARGET_COLUMN].value_counts().sort_index())
```

```python
X = df.drop(columns=[TARGET_COLUMN])
y = df[TARGET_COLUMN]

print("Feature shape:", X.shape)
print("Target shape:", y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=RANDOM_STATE,
    stratify=y
)

print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)
```

```python
train_data = X_train.copy()
train_data[TARGET_COLUMN] = y_train

test_data = X_test.copy()
test_data[TARGET_COLUMN] = y_test

train_data = train_data.sort_index()
test_data = test_data.sort_index()

train_data.to_csv(PROCESSED_DIR / "train_data.csv", index=False)
test_data.to_csv(PROCESSED_DIR / "test_data.csv", index=False)

# Required later for Streamlit upload
test_data.to_csv("../test_data.csv", index=False)

print("Processed datasets saved successfully.")
```

```python
print("Training target distribution:")
display(train_data[TARGET_COLUMN].value_counts().sort_index())

print("Testing target distribution:")
display(test_data[TARGET_COLUMN].value_counts().sort_index())

print("Training file:", PROCESSED_DIR / "train_data.csv")
print("Testing file:", PROCESSED_DIR / "test_data.csv")
```
